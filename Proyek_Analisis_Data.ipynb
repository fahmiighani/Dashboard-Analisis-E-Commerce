**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

### Explore Pertanyaan 1: Kontribusi pendapatan per kategori produk (2017)

```python
# Pesanan delivered pada tahun 2017
orders_2017 = orders_dataset_df[(orders_dataset_df['order_status'] == 'delivered') & (orders_dataset_df['order_purchase_timestamp'].dt.year == 2017)].copy()
print('Jumlah pesanan delivered tahun 2017:', len(orders_2017))

# Gabungkan item pesanan 2017 dengan kategori produk
order_items_2017 = (order_item_df
              .merge(orders_2017[['order_id']], on='order_id', how='inner')
              .merge(products_clean[['product_id', 'category']], on='product_id', how='left'))
order_items_2017['category'] = order_items_2017['category'].fillna('unknown')

total_revenue_2017 = order_items_2017['price'].sum()
print(f'Total revenue 2017 (delivered): {total_revenue_2017:,.2f}')

category_revenue = (order_items_2017.groupby('category')
                    .agg(revenue=('price', 'sum'), jumlah_item=('order_id', 'count'),
                         jumlah_pesanan=('order_id', 'nunique'))
                    .sort_values('revenue', ascending=False))
category_revenue['persentase_revenue'] = category_revenue['revenue'] / total_revenue_2017 * 100
category_revenue['kumulatif_persentase_revenue'] = category_revenue['persentase_revenue'].cumsum()

print('Jumlah kategori:', len(category_revenue))
category_revenue.head(10).round(2)
```

```python
THRESHOLD = 10
above_threshold = category_revenue[category_revenue['persentase_revenue'] > THRESHOLD]
print(f'Kategori dengan kontribusi > {THRESHOLD}% : {len(above_threshold)}')
print(f'Kontribusi tertinggi         : {category_revenue.index[0]} ({category_revenue["persentase_revenue"].iloc[0]:.2f}%)')
print(f'Jumlah kategori >= 5%        : {(category_revenue["persentase_revenue"] >= 5).sum()} '
      f'(total {category_revenue.loc[category_revenue["persentase_revenue"] >= 5, "persentase_revenue"].sum():.1f}% revenue)')
print(f'Kontribusi 5 kategori teratas : {category_revenue["persentase_revenue"].head(5).sum():.1f}%')
print(f'Kontribusi 10 kategori teratas: {category_revenue["persentase_revenue"].head(10).sum():.1f}%')

# Analisis Sensitivitas (sensitivity): apakah kesimpulan berubah jika definisi revenue / filter status diubah?
tmp = order_items_2017.assign(price_freight=order_items_2017['price'] + order_items_2017['freight_value'])
kumulatif_persentase_revenue = tmp.groupby('category')['price_freight'].sum()
kumulatif_persentase_revenue = (kumulatif_persentase_revenue / kumulatif_persentase_revenue.sum() * 100).sort_values(ascending=False)

items_all_2017 = (order_item_df
                  .merge(orders_dataset_df[orders_dataset_df['order_purchase_timestamp'].dt.year == 2017][['order_id']], on='order_id')
                  .merge(products_clean[['product_id', 'category']], on='product_id', how='left'))
items_all_2017['category'] = items_all_2017['category'].fillna('unknown')
share_all = items_all_2017.groupby('category')['price'].sum()
share_all = (share_all / share_all.sum() * 100).sort_values(ascending=False)

print('\nAnalisis sensitivitas:')
print(f'  price + freight (delivered) : {kumulatif_persentase_revenue.index[0]} {kumulatif_persentase_revenue.iloc[0]:.2f}%')
print(f'  price semua status 2017     : {share_all.index[0]} {share_all.iloc[0]:.2f}%')
```

**Insight:** (Opsional)
- Total revenue pesanan delivered tahun 2017 sebesar R$ 5,96 juta yang tersebar di 73 kategori (termasuk `unknown`).
- Tidak ada satu pun kategori yang melampaui 10%. Kategori tertinggi adalah 'bed_bath_table` dengan 8,23% (R$ 490,6 ribu), disusul `watches_gifts` (7,98%) dan `health_beauty` (7,95%).
- Pendapatan tersebar merata: 3 kategori teratas hanya selisih sekitar 0,3 poin persentase, 5 kategori teratas menyumbang 38,0%, dan 10 kategori teratas 63,1%. Terdapat 8 kategori dengan kontribusi minimal 5% yang bersama-sama menyumbang 54,9% revenue.
- Hasil ini kokoh terhadap perubahan definisi: jika revenue dihitung dengan ongkos kirim (8,39%) atau mencakup semua status pesanan (8,10%), tetap tidak ada kategori yang melewati 10%.


### Explore Pertanyaan 2: Keterlambatan pengiriman vs skor ulasan (2017)

```python
# Gabungkan pesanan delivered 2017 dengan ulasan (1 ulasan per pesanan)
delivery_df = (orders_2017
               .merge(reviews_clean[['order_id', 'review_score']], on='order_id', how='left'))

print('Pesanan delivered 2017         :', len(delivery_df))
print('Pesanan tanpa ulasan (dibuang) :', delivery_df['review_score'].isna().sum())
delivery_df = delivery_df.dropna(subset=['review_score']).copy()
print('Pesanan dianalisis             :', len(delivery_df))

# Definisi awal: perbandingan timestamp penuh (delivered > estimated)
delivery_df['is_late'] = delivery_df['order_delivered_customer_date'] > delivery_df['order_estimated_delivery_date']

# Pengecekan: estimasi selalu tersimpan pada pukul 00:00, sedangkan tanggal tiba mencatat jam aktual.
# Pesanan yang tiba pada tanggal estimasi (tetapi lewat pukul 00:00) ikut terhitung terlambat.
estimasi = delivery_df["order_estimated_delivery_date"]
aktual = delivery_df["order_delivered_customer_date"]
tanggal_sama = aktual.dt.normalize().eq(estimasi.dt.normalize())
bukti = delivery_df.loc[
    tanggal_sama & delivery_df["is_late"],
    ["order_estimated_delivery_date", "order_delivered_customer_date", "is_late"],
].copy()
bukti["is_late_tanggal_kalender"] = False
terlambat_timestamp = delivery_df['is_late'].sum()
print(f"Pesanan yang tiba pada tanggal estimasi tetapi ditandai terlambat: {len(bukti)}")
display(bukti.head(5))

# Koreksi: keterlambatan dihitung pada level tanggal kalender (bukan timestamp penuh)
delivery_df['days_late'] = (aktual.dt.normalize() - estimasi.dt.normalize()).dt.days
delivery_df['is_late'] = delivery_df['days_late'] > 0
delivery_df['delivery_status'] = np.where(delivery_df['is_late'], 'Terlambat', 'Tepat waktu')

print(f'Pesanan terlambat (timestamp penuh)  : {terlambat_timestamp:,}')
print(f'Pesanan terlambat (tanggal kalender) : {delivery_df["is_late"].sum():,}')
print(f'Persentase pesanan terlambat         : {delivery_df["is_late"].mean() * 100:.2f}%')
print(f'Rata-rata skor ulasan (semua)        : {delivery_df["review_score"].mean():.3f}')

summary_status = (delivery_df.groupby('delivery_status')['review_score']
                  .agg(jumlah_pesanan='count', rata_rata_skor='mean', median_skor='median', std_skor='std')
                  .round(3))
summary_status
```

```python
# Skor ulasan berdasarkan tingkat keterlambatan (hari kalender melewati estimasi)
bins = [-np.inf, 0, 3, 7, 14, np.inf]
labels = ['Tepat waktu', 'Telat 1-3 hari', 'Telat 4-7 hari', 'Telat 8-14 hari', 'Telat >14 hari']
delivery_df['delay_group'] = pd.cut(delivery_df['days_late'], bins=bins, labels=labels)

summary_delay = (delivery_df.groupby('delay_group', observed=True)['review_score']
                 .agg(jumlah_pesanan='count', rata_rata_skor='mean').round(3))
summary_delay['di_bawah_2.5'] = summary_delay['rata_rata_skor'] < 2.5
display(summary_delay)

# Rincian per hari (1-7 hari) untuk melihat titik tembus ambang 2,5
per_hari = (delivery_df[delivery_df['days_late'].between(1, 7)]
            .groupby('days_late')['review_score']
            .agg(jumlah_pesanan='count', rata_rata_skor='mean').round(3))
per_hari['di_bawah_2.5'] = per_hari['rata_rata_skor'] < 2.5
```

```python
# Distribusi skor ulasan (%): tepat waktu vs terlambat
dist_skor = (pd.crosstab(delivery_df['delivery_status'], delivery_df['review_score'], normalize='index') * 100).round(1)
dist_skor
```

**Insight :**
- Perbandingan keterlambatan dikoreksi sesuai tanggal kalender: 421 pesanan yang tiba pada tanggal estimasi (tetapi lewat pukul 00:00) dihitung tepat waktu.
- Dari 43.098 pesanan delivered 2017 yang memiliki ulasan, 2.384 pesanan (5,5%) terlambat.
- Rata-rata skor pesanan tepat waktu adalah 4,28, sedangkan pesanan terlambat hanya 2,32 dengan median 1.
- Secara keseluruhan rata-rata skor pesanan terlambat (2,32) **sudah di bawah ambang 2,5**.
- Semakin lama terlambat, semakin rendah skor: telat 1-3 hari rata-rata 3,25; 4-7 hari 2,22; 8-14 hari 1,75; dan >14 hari 1,77. Ambang 2,5 bintang tertembus saat keterlambatan melewati 3 hari. Dengan per hari, rata-rata skor mulai di bawah 2,5 pada hari ke-5.

## Analisis Lanjutan (Opsional)

- Pareto / kontribusi kumulatif (Pertanyaan 1): menentukan konsentrasi pendapatan, yaitu berapa kategori yang dibutuhkan untuk mencapai sebagian besar revenue, ketika tidak ada kategori yang melampaui ambang 10%. Hasilnya dipakai untuk membuat tingkatan prioritas alternatif.
- Binning/segmentasi manual (Pertanyaan 2): mengelompokkan lama keterlambatan (1-3, 4-7, 8-14, >14 hari) untuk menemukan titik keterlambatan saat skor mulai di bawah 2,5, karena rata-rata gabungan menyembunyikan perbedaan antar kelompok.
- Uji Mann-Whitney U dan korelasi Spearman (Pertanyaan 2): karena skor ulasan berskala ordinal (1-5) dan tidak berdistribusi normal, uji nonparametrik dipakai untuk menguji apakah perbedaan skor antara pesanan tepat waktu dan terlambat signifikan secara statistik, bukan kebetulan. Uji ini hanya menunjukkan asosiasi, bukan hubungan sebab-akibat.

```python
# Tingkatan prioritas kategori (Pertanyaan 1)
category_revenue['tier'] = pd.cut(category_revenue['persentase_revenue'],
                                  bins=[-np.inf, 2, 5, THRESHOLD, np.inf],
                                  labels=['Tier 4 (<2%)', 'Tier 3 (2-5%)', 'Tier 2 (5-10%)', 'Tier 1 (>10%)'])
tier_summary = (category_revenue.groupby('tier', observed=True)
                .agg(jumlah_kategori=('revenue', 'count'), total_persentase_revenue=('persentase_revenue', 'sum'))
                .round(1).sort_index(ascending=False))
display(tier_summary)
print('Kategori Tier 2 (5-10%):', ', '.join(category_revenue[category_revenue['tier'] == 'Tier 2 (5-10%)'].index))

# Uji asosiasi (Pertanyaan 2): hasil menunjukkan hubungan/asosiasi, bukan sebab-akibat
late_scores = delivery_df.loc[delivery_df['is_late'], 'review_score']
ontime_scores = delivery_df.loc[~delivery_df['is_late'], 'review_score']

u_stat, p_value = stats.mannwhitneyu(late_scores, ontime_scores, alternative='less')
rho, p_rho = stats.spearmanr(delivery_df['days_late'], delivery_df['review_score'])
rho_late, _ = stats.spearmanr(delivery_df.loc[delivery_df['is_late'], 'days_late'], late_scores)

print(f'Mann-Whitney U (terlambat < tepat waktu): U={u_stat:,.0f}, p-value={p_value:.3g}')
print(f'Korelasi Spearman days_late vs skor (semua pesanan)  : {rho:.3f} (p={p_rho:.3g})')
print(f'Korelasi Spearman days_late vs skor (hanya terlambat): {rho_late:.3f}')
print(f'Proporsi ulasan bintang 1-2: tepat waktu {(ontime_scores <= 2).mean() * 100:.1f}% vs terlambat {(late_scores <= 2).mean() * 100:.1f}%')
```

**Insight:** (Opsional)
- Pertanyaan 1: Tanpa kategori Tier 1 (>10%), terdapat 8 kategori Tier 2 (5-10%) yang menyumbang 54,9% revenue sehingga layak dijadikan kelompok prioritas pengganti ambang 10%.
- Pertanyaan 2: Keterlambatan berkaitan dengan skor ulasan yang lebih rendah secara statistik (Mann-Whitney, p mendekati 0). Uji ini hanya menunjukkan asosiasi, bukan hubungan sebab-akibat. Korelasi Spearman terhadap seluruh pesanan lemah (sekitar -0,17) karena 94,5% pesanan tiba tepat waktu. Hubungan lebih terlihat pada kelompok yang terlambat: korelasi Spearman di antara pesanan terlambat sekitar -0,37, dan ulasan bintang 1-2 mencapai 61,5% pada pesanan terlambat dibanding 9,2% pada pesanan tepat waktu.
